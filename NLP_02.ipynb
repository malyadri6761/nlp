**Text Normalizatoin : Transforms raw text into a clean, consistent format before processing.**

---

```python
import re
import string

def normalize_text(text:str)->str:
  """
    Standardizes a text string by lowercse formatting, stripping white space
    and deleting structural punctutaion or numerical artifacts.
  """

  if not text:
      return ""

  text = text.lower()
  text = re.sub(r'[\s\n\r\t]+'," ",text)

  punctuation_table = str.maketrans("","",string.punctuation)
  text = text.translate(punctuation_table)

  return text.strip()

raw_sentence = "   Hello! Welcome to AIM1014. Is this Chapter 1: \n  Mathematics?   "
clean_sentence = normalize_text(raw_sentence)

print("Original:")
print(f"'{raw_sentence}'")
print("\nNormalized:")
print(f"'{clean_sentence}'")
```

**Regular Expression**

---


*   findall - Returns a list containing all matches.
*   search - Returns a Match object if there is a match anywhere in the string.
*    split - Returns a list where the string has been split at each match.
*    sub   - Replaces one or many matches with a string.


---

    Character  
     []  - A set of Charcters
           Ex : "[a-m]"

     \  - Signals a spacial sequcence (can also be used to escapte special characters)
           Ex : "\d"
    
     . - Any character (except newline character)
           Ex : "he..o"
    
     ^ - Starts with
           Ex : "^hello"

     $ - Ends with
           Ex : "planet$"
    
     * - Zero or more occurrences
           Ex : "he.*o"
      
     + - One or more Occurrences
           Ex : "he.+o"
     
     ? - Zero or one occurrences
           Ex : "he.?o"
    
     {} - Exactly the specified number of occurrences
           Ex : "he.{2}.o"
      
      | -  Either or
            Ex : "falls|stays"
      
      () - capture and group

---

**Flag**

      re.ASCII - re.A - Returns only ASCII matches.

      re.DEBUG - Returns debug information.

      re.DOTALL - Makes the .character match all characters(including newline character)

      re.IGNORECASE - re.I - Case-insensitive matching

      re.MULTILINE - re.M - Returns matches at the start/end of each line

      re.NOFLAG - Specifies that no flag is set for this pattern

      re.UNICODE - re.U - Returns Unicode matches. This is default from Python 3. for Python 2: Use this flag to return only unicode matches.

      re.VERBOSE - re.X Allows whitespaces and comments inside patterns. Makes the pattern more readable.

---

**Special Sequences**

    \A	Returns a match if the specified characters are at the beginning of the string	"\AThe"

    \b	Returns a match where the specified characters are at the beginning or at the end of a word
    (the "r" in the beginning is making sure that the string is being treated as a "raw string")	r"\bain"

    r"ain\b"

    \B	Returns a match where the specified characters are present, but NOT at the beginning (or at the end) of a word
    (the "r" in the beginning is making sure that the string is being treated as a "raw string")	r"\Bain"

    r"ain\B"

    \d	Returns a match where the string contains digits (numbers from 0-9)	"\d"
    \D	Returns a match where the string DOES NOT contain digits	"\D"
    \s	Returns a match where the string contains a white space character	"\s"
    \S	Returns a match where the string DOES NOT contain a white space character	"\S"
    \w	Returns a match where the string contains any word characters (characters from a to Z, digits from 0-9, and the underscore _ character)	"\w"
    \W	Returns a match where the string DOES NOT contain any word characters	"\W"
    \Z	Returns a match if the specified characters are at the end of the string

---

**Sets**

    [arn]	Returns a match where one of the specified characters (a, r, or n) is present

    [a-n]	Returns a match for any lower case character, alphabetically between a and n

    [^arn]	Returns a match for any character EXCEPT a, r, and n

    [0123]	Returns a match where any of the specified digits (0, 1, 2, or 3) are present

    [0-9]	Returns a match for any digit between 0 and 9

    [0-5][0-9]	Returns a match for any two-digit numbers from 00 and 59

    [a-zA-Z]	Returns a match for any character alphabetically between a and z, lower case OR upper case

    [+]	In sets, +, *, ., |, (), $,{} has no special meaning, so [+] means: return a match for any + character in the string

```python
# findall() - Function

import re

txt = "The rain in IIT ROPAR"

x = re.findall("IIT",txt)
print(x)

# List contains the matches in the order they are found.
```

```python
# serach() - Function

import re

txt = "Present I am at Library learning Regular Expression."

x = re.search("am",txt)
print(x.start()) # staring index
print(x.end()) # ending index
print(x.span()) # span of the index
```

```python
# split() - Function

import re

txt = " I was decided to learn NLP(Natural Language Processing) and Regular Expression re module helps for learning natural language processing."

x = re.split("\s",txt)
print(x)
```

```python
# split() - Function

import re

txt = " I was decided to learn NLP(Natural Language Processing) and Regular Expression re module helps for learning natural language processing."

# maxsplit - It split upto certain no.of spaces
x = re.split("\s",txt,5)
print(x)
```

```python
# sub() - function replaces the matches with teh text of your choice.

import re
txt = "While adopting new environment it's difficult at initially."

x = re.sub("\s","*",txt)
print(x)
```

```python
# count - Controlling the no.of replacement of the characters.

import re

txt = "While adopting new environment it's difficult at initially."

x = re.sub("\s","*",txt,2)
print(x)
```

```python
# Math Object - an object containing information about the search and the result.

import re

txt = "How are you man?"

x = re.search("are",txt)
print(x)
```

```python
# Math Object Functions

# .span() - returns a tuple containing the start, and end positions of the match.
# .string - returns the string passed into the function.
# .group() - returns the part of the string where there was a match

import re

txt = """ It looks like your code snippet cut off right at the end with reader att. You are likely trying to access an attribute or method of the PdfReader object, or perhaps you want to extract specific metadata.Here are the most common pypdf reader attributes and methods you might be looking for"""

x = re.search("code",txt)

print(x.span())
print(x.string)
print(x.group())
```
