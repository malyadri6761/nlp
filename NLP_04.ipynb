**N-gram**

---

```
  A continuous sequence of N items from a given smaple of text.
  When N=2, it is called a Bigram (pairs of consecutive words).
  Generating bigrams helps machine learning models capture context that single words lose( for example,
  understanding that "machine" and "learning" are tied together as "machine learning").
  

---

**Pure Python Method**

---

````python
text = "Natural Language Processing handles human languages."
words = text.lower().split()

bigrams = list(zip(words,words[1:]))

print(bigrams)
````

---

**NLTK Method (Highle Customisable)**

---

```
  NLTK provides dedicated utility functions (bigrams and ngrams) that accept iterable token list.
  It handles tokenisation cleanly beforehand.

```

````python
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import bigrams
nltk.download('punkt_tab')

text = "Natural language processing handles human languages"

# tokens = [w.lower() for w in word_tokenize(text) if w.isalnum()]

tokens = word_tokenize(text.lower())
nltk_bigrams = list(bigrams(tokens))

print("NLTK Bigrams : ")
print(nltk_bigrams)
````

---

**N-grams using NLTK**

---

````python
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

text = "Large language models are trasforming AI technologies."

tokens = [token.lower() for token in word_tokenize(text) if token.isalnum()]

trigrams = list(ngrams(tokens,3))
print(f"Trigrams : {trigrams}")

fivegrams = list(ngrams(tokens,5))
print(f"Fivegrams : {fivegrams}")
````

**spaCy Method (Context-Aware Bigrams)**

---

```
  spaCy doesn't have a direct bigrams() function out of box, it provides structural objects.
  You can combine its rapid tokenisation pipeline with a list comprehensive to get highly accurate linguistic bigrams.

```

````python
import spacy

nlp = spacy.load('en_core_web_sm')
doc = nlp(text)

clean_tokens = [token.text for token in doc if not token.is_punct]
spacy_bigrams = list(zip(clean_tokens,clean_tokens[1:]))

print("spaCy Bigrams : ")
print(spacy_bigrams)
````

**spaCy Method (Linguistic Filtering)**

---

```
  spaCy allows you to generate N-grams based on clean grammatical data (like
  keeping only root lemmas or skipping stop words).

```



````python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Large language models are transforming AI technologies.")

# Filter out punctuation and keep base forms (lemmas)
clean_lemmas = [token.lemma_.lower() for token in doc if not token.is_punct]

n = 3
spacy_ngrams = [tuple(clean_lemmas[i:i+n]) for i in range(len(clean_lemmas) - n + 1)]

print(f"spaCy Lemma N-grams (N={n}):")
print(spacy_ngrams)
````

**Jaccard Similarity - Similarity between documents**

---

```
  similarity(A,B) = |A intersection B|/|A union B|


---

**spaCy Method**

---

````python
# !pip install pypdf
import spacy
from pypdf import PdfReader
nlp = spacy.load("en_core_web_sm")

path1 = "/content/merged.pdf"
path2 = "/content/tutorial_03.pdf"

reader1 = PdfReader(path1)
reader2 = PdfReader(path2)

set1 = set()
set2 = set()

for index,page in enumerate(reader1.pages):
  text = page.extract_text()
  doc = nlp(text)
  clean_lemmas = [token.lemma_.lower() for token in doc if not token.is_punct]
  set1.update(clean_lemmas)

for index,page in enumerate(reader2.pages):
  text = page.extract_text()
  doc = nlp(text)
  clean_lemmas = [token.lemma_.lower() for token in doc if not token.is_punct]
  set2.update(clean_lemmas)

inter_section = set1.intersection(set2)
union = set1.union(set2)
similarity = len(inter_section)/len(union)
print(f"similarity : {similarity:.4f}")
````

**Cosine Similarity**

---

```

  Measure how similar two vectors are, irrespective of their size.
  In NLP, instead of looking at raw overlapping word sets(like jaccard),it treats documents as structural directions in a geometric vector space.
  it calculates the cosine of the angle between two text vectors.
      Formula :     
          Cosine Similarity(A,B) = A.B/||A|| ||B||

  1.0 : The documents point in the exact same vector directoin(semantically identical).

  0.0 : The documents are completely orthogonal(share absolutely zero words/context)
```


---

**Pure Python Method(Count Vectorizer Matrix)**

---

````python
import math
import nltk
from pypdf import PdfReader
from nltk.tokenize import word_tokenize

reader1 = PdfReader(path1)
reader2 = PdfReader(path2)

words1 = []
words2 = []




# Adding tokens to words1
for index,pages in enumerate(reader1.pages):
  text = pages.extract_text()
  tokens = [token.lower() for token in word_tokenize(text) if token.isalnum()]
  words1.extend(tokens)

# Adding tokens to words2
for index,pages in enumerate(reader2.pages):
  text = pages.extract_text()
  tokens = [token.lower() for token in word_tokenize(text) if token.isalnum()]
  words2.extend(tokens)

vocabulary = sorted(list(set(words1 + words2)))

# Build term frequency vectors
vector1 = [words1.count(word) for word in vocabulary]
vector2 = [words2.count(word) for word in vocabulary]


# Dot product
dot_product = sum(w1*w2 for w1,w2 in zip(vector1,vector2))
magnitude1 = math.sqrt(sum(w**2 for w in vector1))
magnitude2 = math.sqrt(sum(w**2 for w in vector2))

# Cosine Similarity
cosine_similarity = dot_product/(magnitude1*magnitude2)
print(f"Cosine Similarity : {cosine_similarity:.4f}")

if magnitude1 == 0 or magnitude2 == 0:
    similarity = 0.0
else:
    similarity = dot_product / (magnitude1 * magnitude2)

print(f"Pure Python Cosine Similarity: {similarity:.4f}")


````

**scikit-learn Method(Industry Standard for Keyword counts)**

---

```
  Using TfidfVectorizer automatically scales down common filter words (like "is" or
  "the") and scales up rare meaningful temrs. It calculates the structural similarity matrix instantly.
```


````python
# Run first: pip install scikit-learn
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Natural language processing handles human language text.",
    "Large language models handle deep learning processing tasks."
]

# 1. Compute TF-IDF matrix
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

# 2. Compute similarity between document 0 and document 1
similarity_matrix = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])
similarity_score = similarity_matrix[0][0]

print(f"Scikit-Learn TF-IDF Cosine Similarity: {similarity_score:.4f}")
````

**spaCy Method (True deep Learning Semantic similarity)**

---

```
  Standard keyowrd cosine similarity will give a 0.0 score if two documents don't share
  exact matching words. spaCy avoids this by using Word Vectors(Embeddings). It
  can determine that "king" and "queen" are close geometrically even if they are typed
  differently.

  Note: You must use the medium (_md) or large (_lg) spaCy models, as the small (_sm) models do not include real word vector weights.


```

````python
# Run first: pip install spacy && python -m spacy download en_core_web_md
import spacy

# Load model with pre-trained word vector embeddings
nlp = spacy.load("en_core_web_sm")

doc1 = nlp("The weather today is completely freezing and snowy.")
doc2 = nlp("It is an incredibly cold winter morning outside.")

# spaCy uses an internal vector dot-product calculation automatically
semantic_similarity = doc1.similarity(doc2)

print(f"spaCy Semantic Embedding Similarity: {semantic_similarity:.4f}")
````
