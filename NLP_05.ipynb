**Bag Of Words**

---


```python
# For single file

import nltk
import re
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize

text = """I am learning nlp.
          I am learning machine learning"""

text = re.sub(r'[^\w\s]','',text)
tokens = word_tokenize(text.lower())

unique  = list(set(tokens))

bag_of_word = []
for word in unique:
  bag_of_word.append(tokens.count(word))
print(bag_of_word)

```

```python
import re
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

tokens = []
vocabulary = set()

documents = ["I am Learning NLP","I am learning Maching leanring","I love aritificial intelligence"]

# Reading each string from documents and adding to tokens into tokens and vocabulary
for string in documents:
  # ^ negation
  # ^\w\s not a word character or not a space
  text_cleaned = re.sub(r'[^\w\s]','',string)
  token = word_tokenize(text_cleaned.lower())
  tokens.append(token)
  vocabulary.update(token)

unique_vocab = sorted(list(vocabulary))

print("Global vocabulary columns : ")
print(unique_vocab)


# Generate a separate Bag of Words vector for each file
bag_of_words_matrix = []
for token in tokens:
  file_vector = []
  for word in unique_vocab:
    file_vector.append(token.count(word))
  bag_of_words_matrix.append(file_vector)

print(bag_of_words_matrix)
print(tokens)
```
