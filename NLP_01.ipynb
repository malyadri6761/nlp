**(V) Extract Images from a PDF**

----
```
Scans a PDF page by page, identifies all embedded high-quality images, and saves them directly to your directory.
```


````python
import os
from pypdf import PdfReader

pdf_path = "/content/pytorch_.pdf"
output_dir = "/content/output"

os.makedirs(output_dir,exist_ok=True)

reader = PdfReader(pdf_path)

image_count = 0

for page_index,page in enumerate(reader.pages):
  for image_file_object in page.images:
    image_count+=1

    image_name = f'page_{page_index+1}_imge_{image_count}_{image_file_object.name}'
    image_path = os.path.join(output_dir,image_name)

    with open(image_path,"wb") as f:
      f.write(image_file_object.data)

print(f"Done! Extracted {image_count} images to {output_dir}")
````

**Task 3 : Count wrods in a document**

---



````python
from pypdf import PdfReader

pdf_path = "/content/Tutorial_3.pdf"
reader = PdfReader(pdf_path)


total_words_count = 0

for page_index,page in enumerate(reader.pages):
  text = page.extract_text()
  if text:
    words = text.split()
    page_word_count = len(words)
    total_words_count +=page_word_count

print(total_words_count)
````
