**Stemming vs Lemmatization**

---

```
 Stemming : A fast, crude algorithm that chops off the ends of words.
 it often creates non-words( Eg : "studying" becomes "studi").

 ```

 ```
  Lemmatization : A slower, smarter dictionary lookup that ensures that root
  wor morphologically correct.

```

````python
import nltk
nltk.download('wordnet')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

text = "I am present studing natural language processing for building large language model"
word_token = word_tokenize(text.lower())


stemmed_words = [stemmer.stem(word) for word in word_token]
Lemmatize_words = [lemmatizer.lemmatize(word) for word in word_token]

print(f"Stemmed Words : {stemmed_words}")
print(f"Lemmatize_words : {Lemmatize_words}")
````

````python
from pypdf import PdfReader

import nltk
from nltk.tokenize import word_tokenize

path = "/content/tutorial_03.pdf"
reader = PdfReader(path)

word_freq = dict({})

for index,page in enumerate(reader.pages):

  word_token = word_tokenize(page.extract_text().lower())
  for token in word_token:
    word_freq[token] = word_freq.get(token,0)+1

print(word_freq)
print(len(word_freq))
````

**Character Frequency**

````python
def charFreq(string,char_freq):
  for ch in string:
    char_freq[ch] = char_freq.get(ch,0)+1


from pypdf import PdfReader

import nltk
from nltk.tokenize import word_tokenize

path = "/content/tutorial_03.pdf"
reader = PdfReader(path)

char_freq = dict({})

for index,page in enumerate(reader.pages):

  word_token = word_tokenize(page.extract_text().lower())
  for token in word_token:
    charFreq(token,char_freq)

print(char_freq)
print(len(char_freq))
````

**Building Vocabulary using nltk**

---

```
  Best for custome thresholds.
  In real-world applications, you often want to exclude rare words(eg. words that only
  appear once) or common punctuation. NLTK's FreqDist allows you to sort by frequency and filter out tokens based on occurrence.

```

````python
import nltk

from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

nltk.download('punkt')

corpus = [
    "Natural language processing is a subset of AI.",
    "Large language models are built using natural language processing."
]


# 1. Tokenize and clean text
all_tokens = []
for text in corpus:
  tokens = word_tokenize(text.lower())
  cleaned_tokens = [t for t in tokens if t.isalnum()]
  all_tokens.extend(cleaned_tokens)


# 2. Get frequency counts
freq_dist = FreqDist(all_tokens)

# 3. Build vocab (Example : only keep words that appear at least 1 time)
min_freq = 3
unique_vocab = [word for word,freq in freq_dist.items() if freq>= min_freq]


# 4.Map to IDs (reserve index 0 for unknown words '<UNK>')
vocab_with_ids = {'<UNK>':0}
for idx,word in enumerate(sorted(unique_vocab),start=1):
  vocab_with_ids[word] = idx

print("NLTK Vocabulary : \n",vocab_with_ids)
````

**spaCy Method (Best for Modern NLP Pipelines)**

---

```
  When you process text in spaCy, it maintains an internal string store (nlp.vocab.strings).
  This automatically assigns a 64-bit hash integer to every unique string across the entire lifetime of the pipeline, preventing memory bloat.

```

````python
import spacy

nlp = spacy.load("en_core_web_sm")

corpus = [
    "Natural language processing is a subset of AI.",
    "Large language models are built using natural language processing."
]

unique_words = set()

# Process documents and extract unique base lemmas(roots)
for text in corpus:
  doc = nlp(text.lower())
  for token in doc:
    if not token.is_punct and not token.is_space:
      unique_words.add(token.lemma_)

spacy_vocab = {lemma: idx for idx,lemma in enumerate(sorted(unique_words))}
print("spaCy Lemma Vocabulary : \n",spacy_vocab)
````


| Metric | Pure Python | NLTK | spaCy |
| :--- | :--- | :--- | :--- |
| **Token Handling** | Raw text splitting | Linguistic token rules | Advanced linguistic parsing |
| **Vocabulary Basis** | Exact words as typed | Exact words as typed | Can build using base **Lemmas** |
| **Punctuation Cleanliness** | Manual handling required | High control via string filtering | Native boolean flags (`token.is_punct`) |
| **Best Feature** | No installation required | Frequency threshold filtering | Memory-optimized global string hash |
